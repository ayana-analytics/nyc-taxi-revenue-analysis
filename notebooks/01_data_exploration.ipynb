## 6. Feature Engineering

New variables are created to support time-based, revenue and customer behavior analysis.

```python
df_clean["trip_duration_min"] = (
    df_clean["tpep_dropoff_datetime"]
    - df_clean["tpep_pickup_datetime"]
).dt.total_seconds() / 60
```

```python
df_clean["trip_duration_min"].describe()
```

```python
df_clean = df_clean[
    (df_clean["tpep_pickup_datetime"] >= "2026-01-01")
    & (df_clean["tpep_pickup_datetime"] < "2026-02-01")
]
```

```python
df_clean = df_clean[
    (df_clean["trip_distance"] > 0)
    & (df_clean["fare_amount"] > 0)
    & (df_clean["total_amount"] > 0)
    & (df_clean["tip_amount"] >= 0)
    & (df_clean["trip_duration_min"] > 0)
]
```

```python
print("Original dataset:", len(df))
print("Clean dataset:", len(df_clean))

removed_rows = len(df) - len(df_clean)
removed_percent = removed_rows / len(df) * 100

print("Rows removed:", removed_rows)
print("Percentage removed:", round(removed_percent, 2), "%")
```

```python
df_clean["pickup_hour"] = (
    df_clean["tpep_pickup_datetime"].dt.hour
)

df_clean["day_of_week"] = (
    df_clean["tpep_pickup_datetime"].dt.day_name()
)

df_clean["is_weekend"] = (
    df_clean["tpep_pickup_datetime"]
    .dt.dayofweek
    .isin([5, 6])
)
```

```python
df_clean["tip_percentage"] = (
    df_clean["tip_amount"]
    / df_clean["fare_amount"]
    * 100
)
```

## 10. Payment Type Analysis

This section compares trip volume and average payment amounts across payment methods.
According to the NYC TLC data dictionary, cash tips are not recorded in the `tip_amount` field, so tipping behavior is not compared across payment types.

```python
payment_labels = {
    0: "Flex Fare",
    1: "Credit card",
    2: "Cash",
    3: "No charge",
    4: "Dispute",
    5: "Unknown",
    6: "Voided trip"
}

df_analysis["payment_method"] = (
    df_analysis["payment_type"].map(payment_labels)
)
```

```python
payment_summary = (
    df_analysis.groupby("payment_method")
    .agg(
        trips=("payment_type", "size"),
        avg_trip_amount=("total_amount", "mean"),
        total_payments=("total_amount", "sum")
    )
    .sort_values("trips", ascending=False)
)

payment_summary
```

```python
payment_summary["trips"].plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("NYC Yellow Taxi Trips by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Number of Trips")
plt.xticks(rotation=45)
plt.show()
```

## 11. Airport Trip Analysis

Airport-related trips are analyzed as a separate customer segment because they may differ in distance, duration and payment value from regular taxi trips.

```python
df_analysis["is_airport_trip"] = (
    (df_analysis["Airport_fee"] > 0)
    | (df_analysis["RatecodeID"] == 2)
)
```

```python
airport_summary = (
    df_analysis.groupby("is_airport_trip")
    .agg(
        trips=("is_airport_trip", "size"),
        avg_distance=("trip_distance", "mean"),
        avg_duration=("trip_duration_min", "mean"),
        avg_trip_amount=("total_amount", "mean")
    )
)

airport_summary
```

```python
airport_labels = {
    False: "Non-airport",
    True: "Airport"
}

airport_plot = airport_summary.copy()
airport_plot.index = airport_plot.index.map(airport_labels)

airport_plot["avg_trip_amount"].plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Average Trip Amount: Airport vs Non-Airport")
plt.xlabel("Trip Segment")
plt.ylabel("Average Trip Amount ($)")
plt.xticks(rotation=0)
plt.show()
```

## 12. SQL Analysis

SQL is used to query the cleaned taxi trip data and combine it with NYC taxi zone information using SELECT, WHERE, GROUP BY, ORDER BY and JOIN.

```python
import sqlite3
```

```python
zones = pd.read_csv("../data/taxi_zone_lookup.csv")

zones.head()
```

```python
conn = sqlite3.connect("../data/nyc_taxi.db")
```

```python
sql_trips = df_analysis[
    [
        "tpep_pickup_datetime",
        "PULocationID",
        "DOLocationID",
        "trip_distance",
        "trip_duration_min",
        "payment_type",
        "payment_method",
        "fare_amount",
        "tip_amount",
        "total_amount",
        "cbd_congestion_fee",
        "Airport_fee",
        "pickup_hour",
        "day_of_week",
        "is_airport_trip"
    ]
].copy()
```

```python
sql_trips.to_sql(
    "trips",
    conn,
    if_exists="replace",
    index=False
)

zones.to_sql(
    "zones",
    conn,
    if_exists="replace",
    index=False
)
```

```python
pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)
```

### SQL Query 1 — Preview Trips

```python
query = """
SELECT *
FROM trips
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```

### SQL Query 2 — High-Value Trips

```python
query = """
SELECT
    trip_distance,
    fare_amount,
    tip_amount,
    total_amount
FROM trips
WHERE total_amount > 100
ORDER BY total_amount DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

### SQL Query 3 — Payment Method Summary

```python
query = """
SELECT
    payment_method,
    COUNT(*) AS trips,
    ROUND(AVG(total_amount), 2) AS avg_trip_amount,
    ROUND(SUM(total_amount), 2) AS total_payments
FROM trips
GROUP BY payment_method
ORDER BY trips DESC;
"""

pd.read_sql_query(query, conn)
```

### SQL Query 4 — Evening Trip Analysis

```python
query = """
SELECT
    pickup_hour,
    COUNT(*) AS trips,
    ROUND(AVG(total_amount), 2) AS avg_trip_amount
FROM trips
WHERE pickup_hour BETWEEN 16 AND 20
GROUP BY pickup_hour
ORDER BY pickup_hour;
"""

pd.read_sql_query(query, conn)
```

### SQL Query 5 — Top Pickup Zones by Trip Volume

```python
query = """
SELECT
    z.Borough,
    z.Zone,
    COUNT(*) AS trips,
    ROUND(SUM(t.total_amount), 2) AS total_payments
FROM trips AS t
JOIN zones AS z
    ON t.PULocationID = z.LocationID
GROUP BY
    z.Borough,
    z.Zone
ORDER BY trips DESC
LIMIT 10;
"""

top_pickup_zones = pd.read_sql_query(query, conn)

top_pickup_zones
```

### SQL Query 6 — Top Pickup Zones by Total Payments

```python
query = """
SELECT
    z.Borough,
    z.Zone,
    COUNT(*) AS trips,
    ROUND(AVG(t.total_amount), 2) AS avg_trip_amount,
    ROUND(SUM(t.total_amount), 2) AS total_payments
FROM trips AS t
JOIN zones AS z
    ON t.PULocationID = z.LocationID
GROUP BY
    z.Borough,
    z.Zone
ORDER BY total_payments DESC
LIMIT 10;
"""

top_payment_zones = pd.read_sql_query(query, conn)

top_payment_zones
```
